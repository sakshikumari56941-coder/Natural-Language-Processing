```python
#1.creating and writing to a text file
# open file in write mode
file = open("student.text","w")

file.write("Rahul/n")
file.write("Priya/n")
file.write("Amit/n")
file.write("Neha/n")
file.close()
print("file created succerssfully")
```

```python
#2. writing multiple lines using wrilines()
student = ["R̥ahul/n","Priya/n","Amit/n","Neha/n"]
with open("student.text","w") as file:
    file.writelines(student)
print("data writter succerssfull")

```

```python
#3. Reading the Entire file
with open("student.text","r") as file:
    data = file.read()
    print(data)
```

```python
#4. Reading file line by line
with open("student.text","r") as file:
    for line in file:
        print(line.strip())
```

```python
#5. using readline()
with open("student.text","r") as file:
    line1 = file.readline()
    line2 = file.readline()
print("first line:",line1.strip())
print("second line:",line2.strip())
```

```python
#6. Using Readline
with open("student.text","r") as file:
  lines = file.readlines()
  print(lines)
```

```python
#8. wruting user input to a file
name = input("Enter your name")
age = input("Enter your age")
with open('student.text','a') as file:
    file.write("Name " + name +'/n')
    file.write("Age " + age +'/n')
print("data written successfully")
```

```python
#9. Counting Number of Liner
with open("student.text","r") as file:
    lines = file.readlines()
print('Number of lines:',len(lines))
```

```python
#10. Counting Words in Text File
with open("student.text","r") as file:
  data = file.read()
words = data.split()
print("Number of words:",len(words))
```

```python
#11. counting Characters
with open("student.text","r") as file:
  data = file.read()
print("Number of characters:",len(data))
```

```python
#13. Counting Occurrences of a word
search_word = input("Enter word:")
with open("student.text","r") as file:
  data = file.read()
count = data.lower().count(search_word.lower())
print(" Occurrences:",count)
```

```python
#14. Copyin Content from one file to Another
with open("student.text","r") as source:
    data = source.read()
with open("students.text","w") as destination:
    destination.write(data)
print("data copied succerssfully")
```
