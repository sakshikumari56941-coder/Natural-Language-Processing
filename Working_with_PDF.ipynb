```python
# 2. Read a PDF File Using Python

from PyPDF2 import PdfReader

# Note: Changed path from .txt to .pdf as PdfReader only supports PDF files
reader = PdfReader("/content/Pollution.pdf")

print("Number of pages:", len(reader.pages))

# Read first page
page = reader.pages[0]

text = page.extract_text()

print(text)
```

```python
# Read All pages from a PDF

from PyPDF2 import PdfReader

reader = PdfReader("/content/Pollution.pdf")

for page_number, page in enumerate(reader.pages, start = 1):

  print(f"\n--- Page {page_number}---")

  text = page.extract_text()

  print(text)
```

```python
# 4. Read a  Specific page

from PyPDF2 import PdfReader

reader = PdfReader("/content/Pollution.pdf")

page = reader.pages[2]

print(page.extract_text())
```

```python
# 5. Search for a Word in a PDF

from PyPDF2 import PdfReader

reader = PdfReader("/content/Pollution.pdf")

search_word = "Noise"

for page_number, page in enumerate(reader.pages, start = 1):

  text = page.extract_text()

  if search_word.lower() in text.lower():
    print(f"'{search_word}' found on page {page_number}")
```

```python
# 6. Count Words in a PDF

from PyPDF2 import PdfReader

reader = PdfReader("/content/Pollution.pdf")

word_count = 0

for page in reader.pages:

  text = page.extract_text()

  if text:
    word_count += len(text.split())

print("Total words:", word_count)
```

```python
# 7.  Save Extracted Text into a txt file

from PyPDF2 import PdfReader

reader = PdfReader("/content/Pollution.pdf")

with open("pdf_text.txt", "w", encoding="utf-8") as file:
  for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text()

    file.write(f"\n--- Page{page_number} ---\n")
    file.write(text if text else "")

print("PDF text saved successfully.")
```

```python
# Extract text from all pages and save to a new .txt file
try:
    from PyPDF2 import PdfReader
except ModuleNotFoundError:
    import sys
    import subprocess
    subprocess.check_call([sys.executable, "-m", "pip", "install", "PyPDF2"])
    from PyPDF2 import PdfReader

# Open the PDF
reader = PdfReader('/content/Pollution.pdf')
output_filename = 'extracted_text_all_pages.txt'

# Extract and write text
with open(output_filename, 'w', encoding='utf-8') as f:
    for i, page in enumerate(reader.pages, start=1):
        text = page.extract_text() or ''
        f.write(f'--- Page {i} ---\n')
        f.write(text)
        f.write('\n\n')

print(f'Text successfully extracted and saved to {output_filename}')
```

```python
# 8. Copy the first page into another PDF

from PyPDF2 import PdfReader, PdfWriter

# Read original PDF
reader = PdfReader("/content/Pollution.pdf")

# Create new PDF
writer = PdfWriter()

# Copy first page
writer.add_page(reader.pages[0])

# Save as another PDF
with open("first_page.pdf", "wb") as output:

  writer.write(output)

print("First page copied successfully!")
```
