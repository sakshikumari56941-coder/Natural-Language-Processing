```python
## re.search()

## re.search() searches for a pattern anywhere in the string

import re

text = "Python is a programming language."

result = re.search("Python", text)

if result:
  print("Pattern found")
else:
  print("Pattern not fouund")
```

```python
## re.findall()

## findall() returns all occurences of a pattern

import re

text = "I have 10 apples, 20 oranges and 30 bananas."

numbers = re.search(r"\d+", text)

print(numbers)
```

```python
import re

text = "I have 10 apples, 20 oranges and 30 bananas."

numbers = re.findall(r"\d+", text)

print(numbers)
```

```python
## Extract Email address

import re

text = """

Contact Rahul at rahul@gmail.com
or Neha at neha123@yahoo.com
"""

emails = re.findall(r"[a-zA-z0-9._]+@[a-zA-z0-9]+\.[a-zA-Z]{2,}",text)

print(emails)
```

```python
## Extract Phone numbers

import re

text = "Contact me at 9876543210 or 9123456789."

phones = re.findall(r"\b\d{10}\b", text)

print(phones)
```

```python
# Remove Punctuation

import re

text = "Hello, World! Python is amazing."

clean_text = re.sub(r"[^\w\s]", "", text)

print(clean_text)
```

```python
# Remove extra spaces

import re

text = "python    is    very    useful."

clean_text = re.sub(r"\s+", " ", text)

print(clean_text)
```

```python
# Extract Hashtags

import re

text = "I love #python and #DataScience. #MachineLearning is interesting."

hastags = re.findall(r"#\w+", text)

print(hastags)
```

```python
# Extract Mentions

import re

text = "Thanks @Rahul and @Neha for attending the session."

mentions = re.findall(r"@\w+", text)

print(mentions)
```

```python
# Extract URLs

import re

text = """

visit = https://www.python.org

or https://www.google.com for more information.
"""

urls = re.findall(r"https?://\S+", text)

print(urls)
```

```python
# Extract Words Starting with a Specific Letter

import re

text = "Python programming provides stressful possibilities."

words = re.findall(r"\b[s]\w*", text)

print(words)
```

```python
# Replace Words Using Regex

import re

text = "Python is easy. Python is powerful."

new_text = re.sub(r"Python", "Java", text)

print(new_text)
```

```python
# Case-Insensitive Matching

import re

text = "Python is easy. PYTHON is powerful. python is popular."

matches = re.findall(r"python", text, re.IGNORECASE)

print(matches)
```

```python
# Extract Dates

import re

text = "The meeting is scheduled on 26-08-2026."

date = re.findall(r"\b\d{2}-\d{2}-\d{4}\b", text)

print(date)
```
