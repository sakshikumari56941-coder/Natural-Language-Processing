```python
# then import spacy
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = "python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
#3. Tokenization with Token Index
# Each token has a index indicating its position in the document.
# token.i given the psition?index of the token
import spacy
nlp = spacy.load("en_core_web_sm")
text ="python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.i,token.text)
```

```python
# tokenization with token attributes
# spacy provides several useful attributes for each token.

doc = nlp(text)
for token in doc:
  print(token.text,
        token.is_alpha,
        token.is_punct,
        token.is_space)
```

```python
# Tokenization of a sentence with punctuation
text ="Hello , how are you?"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# tokenization of numbers
text ="I brounght 5 books $100"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# Tokenization of email aaddresses
text ="Contact me at sakshi@example.com"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 8 tokenization of URLs
text ="Visit https://www.example.com for more information"
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# Tokenization of contractions
text ="I'm like python."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
#10. tokenization of Hyphenated words
text ="Data-driven decision-marking is important."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# Removing punctuation after tokenization
text ="python is powerful, easy, and popular!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# Getting only alphabetic token
# we can use is_alpha to select only alphabetic token.
text ="python 3 is easy to learn!"
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text)
```

```python
# converting token into a list
text ="python is popular."
doc = nlp(text)
token = [token.text for token in doc]
print(token)
```

```python
# tokenization of multiple senstences
text ="python is easy to learn. python is powerful."
doc = nlp(text)
for token in doc:
  print(token.text)
# spacy processses the entire text as a doc object
```

```python
# tokenzition and sentenc segmentation
# spacy can also identify individual sentences.
text ="python is eary to learn.It is popular in data scince"
doc = nlp(text)
for sentenc in doc.sents:
  print("sentenc:",sentenc.text)

  for token in sentenc:
    print(token.text)
  print()
```
